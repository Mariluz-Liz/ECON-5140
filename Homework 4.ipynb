**Context:** A retailer rolled out same-day delivery in **some** stores (`treated=1`) in **2024**, but not others (`treated=0`). You have **two** years per store: **2023** (pre) and **2024** (post). Sales per store-year are simulated with a **true ATT** of **+5** (thousand dollars) on treated stores in the post year, on top of store fixed differences and a common time shock.

**Variables in `df_did`:**
- `store_id`: store identifier
- `year`: 2023 or 2024
- `treated`: 1 if the store received same-day delivery
- `post`: 1 if year == 2024
- `Y`: sales (thousands)

**Your tasks:**

1. Compute **cell means** $\bar{Y}$ for the four groups (treated/control $\times$ pre/post) and the **manual DID**: $(\bar{Y}_{T,post}-\bar{Y}_{T,pre})-(\bar{Y}_{C,post}-\bar{Y}_{C,pre})$.

2. Estimate the **saturated regression** $Y \sim \text{treated} + \text{post} + \text{treated} \times \text{post}$ (include a constant if you prefer an equivalent reparameterization — be consistent). Report $\hat{\beta}$ on the interaction and verify it matches the manual DID up to rounding.

3. **Plot** average $Y$ by `year` for treated vs control (two lines). Label axes and add a short title.

4. In 2–3 sentences, relate your plot to the parallel-trends idea (what would *violation* look like in the pre period if you had more pre years?).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import warnings
import seaborn as sns


warnings.filterwarnings("ignore")
rng = np.random.default_rng(5141)

n_treated = 40
n_control = 60
TRUE_ATT = 5.0

rows = []
for i in range(n_treated):
    sid = f"T{i}"
    alpha = rng.normal(12, 2)  # store fixed effect (treated stores differ in levels)
    for year, post in [(2023, 0), (2024, 1)]:
        delta_year = 1.5 if post else 0.0  # common time trend / shock
        treat_effect = TRUE_ATT if post else 0.0
        eps = rng.normal(0, 1.2)
        Y = alpha + delta_year + treat_effect + eps
        rows.append({"store_id": sid, "year": year, "treated": 1, "post": post, "Y": Y})

for j in range(n_control):
    sid = f"C{j}"
    alpha = rng.normal(10, 2)
    for year, post in [(2023, 0), (2024, 1)]:
        delta_year = 1.5 if post else 0.0
        eps = rng.normal(0, 1.2)
        Y = alpha + delta_year + eps
        rows.append({"store_id": sid, "year": year, "treated": 0, "post": post, "Y": Y})

df_did = pd.DataFrame(rows)
print("Simulated store-year panel (2 periods)")
print(f"  True ATT on treated in post: +{TRUE_ATT}")
print(df_did.head(8))
```

Compute cell means for the four groups (treated/control  pre/post) and the manual DID:

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# First, let's examine your data structure
print("Data structure:")
print(df_did.head())
print("\nUnique values in 'treated' column:", df_did['treated'].unique())
print("Unique values in 'post' column:", df_did['post'].unique())

# Let's create the cell means manually without using unstack
# Calculate mean for each combination of treated and post
mean_values = {}
for treat in df_did['treated'].unique():
    for p in df_did['post'].unique():
        subset = df_did[(df_did['treated'] == treat) & (df_did['post'] == p)]
        mean_values[(treat, p)] = subset['Y'].mean()

print("\nManually calculated means:")
for key, value in mean_values.items():
    print(f"treated={key[0]}, post={key[1]}: {value:.4f}")

# Create a DataFrame for the cell means
data = {
    'Pre (2023)': [mean_values.get((0, 0), np.nan), mean_values.get((1, 0), np.nan)],
    'Post (2024)': [mean_values.get((0, 1), np.nan), mean_values.get((1, 1), np.nan)]
}

cell_means_df = pd.DataFrame(data, index=['Control', 'Treated'])
cell_means_df['Difference (Post - Pre)'] = cell_means_df['Post (2024)'] - cell_means_df['Pre (2023)']
cell_means_df.index.name = 'Group'

# Calculate DiD manually
treated_diff = mean_values.get((1, 1), 0) - mean_values.get((1, 0), 0)
control_diff = mean_values.get((0, 1), 0) - mean_values.get((0, 0), 0)
manual_did = treated_diff - control_diff

# Display the results
print("\nFour Cell Means:")
print(cell_means_df.round(4))
print("\nManual DiD Calculation:")
print(f"Treated Difference (Post - Pre): {treated_diff:.4f}")
print(f"Control Difference (Post - Pre): {control_diff:.4f}")
print(f"DiD Estimate (Treated Diff - Control Diff): {manual_did:.4f}")

# If TRUE_ATT is defined
TRUE_ATT = 5.0  # Based on your previous output
print(f"True ATT: {TRUE_ATT}")

# Visualize the DiD if we have valid means
if not np.isnan(manual_did):
    plt.figure(figsize=(10, 6))
    sns.pointplot(data=df_did, x='post', y='Y', hue='treated', 
                markers=['o', 's'], linestyles=['-', '--'], dodge=0.3)
    plt.xticks([0, 1], ['Pre (2023)', 'Post (2024)'])
    plt.ylabel('Average Outcome (Y)')
    plt.xlabel('Time Period')
    plt.title('Difference-in-Differences Visualization')
    plt.legend(title='Group', labels=['Control', 'Treated'])
    plt.grid(True, alpha=0.3)
    
    # Add annotations if we have valid means
    if not np.isnan(treated_diff):
        plt.annotate(f"Diff: {treated_diff:.2f}", 
                    xy=(1, mean_values.get((1, 1), 0)),
                    xytext=(1.05, mean_values.get((1, 1), 0)),
                    fontsize=10)
    
    if not np.isnan(control_diff):
        plt.annotate(f"Diff: {control_diff:.2f}", 
                    xy=(1, mean_values.get((0, 1), 0)),
                    xytext=(1.05, mean_values.get((0, 1), 0)),
                    fontsize=10)
    
    if not np.isnan(manual_did):
        mid_y = (mean_values.get((1, 0), 0) + mean_values.get((1, 1), 0))/2
        plt.annotate(f"DiD: {manual_did:.2f}", 
                    xy=(0.5, mid_y),
                    xytext=(0.5, mid_y + 1),
                    fontsize=12, fontweight='bold',
                    arrowprops=dict(arrowstyle='->'))
    
    plt.show()
else:
    print("\nCannot create visualization due to missing values in cell means.")
```

Estimate the saturated regression  (include a constant if you prefer an equivalent reparameterization — be consistent). Report 
 on the interaction and verify it matches the manual DID up to rounding.

```python
import statsmodels.api as sm


# Using the data that's already been created (df_did)

# Create interaction term
df_did['treated_post'] = df_did['treated'] * df_did['post']

# Estimate the saturated regression model
X = sm.add_constant(df_did[['treated', 'post', 'treated_post']])
model = sm.OLS(df_did['Y'], X)
results = model.fit()

# Print regression results
print("\nSaturated Regression Results:")
print(results.summary().tables[1])

# Extract the coefficients for interpretation
constant = results.params['const']
treated_effect = results.params['treated']
post_effect = results.params['post']
interaction_effect = results.params['treated_post']

print("\nInterpretation of Coefficients:")
print(f"Constant (β₀): {constant:.4f} - Average outcome for control group in pre-period")
print(f"Treated (β₁): {treated_effect:.4f} - Baseline difference between treatment and control groups")
print(f"Post (β₂): {post_effect:.4f} - Time trend common to both groups")
print(f"Treated × Post (β₃): {interaction_effect:.4f} - Difference-in-differences estimate (ATT)")

# Calculate manual DID estimate
mean_treated_post = df_did.loc[(df_did['treated'] == 1) & (df_did['post'] == 1), 'Y'].mean()
mean_treated_pre = df_did.loc[(df_did['treated'] == 1) & (df_did['post'] == 0), 'Y'].mean()
mean_control_post = df_did.loc[(df_did['treated'] == 0) & (df_did['post'] == 1), 'Y'].mean()
mean_control_pre = df_did.loc[(df_did['treated'] == 0) & (df_did['post'] == 0), 'Y'].mean()

manual_did = (mean_treated_post - mean_treated_pre) - (mean_control_post - mean_control_pre)

print("\nComparison with Manual DID Calculation:")
print(f"Manual DID estimate: {manual_did:.4f}")
print(f"Regression DID estimate (treated_post coefficient): {interaction_effect:.4f}")
print(f"Difference between estimates: {abs(manual_did - interaction_effect):.8f}")
print(f"True ATT from simulation: {TRUE_ATT}")

# Visualize the DID
groups = ['Control Pre', 'Control Post', 'Treated Pre', 'Treated Post']
means = [mean_control_pre, mean_control_post, mean_treated_pre, mean_treated_post]

plt.figure(figsize=(10, 6))
bars = plt.bar(groups, means, color=['lightblue', 'lightblue', 'lightgreen', 'lightgreen'])
plt.ylabel('Outcome (Y)')
plt.title('Difference-in-Differences Visualization')

# Add horizontal lines to show the parallel trends assumption
plt.plot([0, 1], [mean_control_pre, mean_control_post], 'b-', linewidth=2)
plt.plot([2, 3], [mean_treated_pre, mean_treated_pre + (mean_control_post - mean_control_pre)], 
         'r--', linewidth=2, label='Counterfactual (no treatment)')
plt.plot([2, 3], [mean_treated_pre, mean_treated_post], 'g-', linewidth=2)

# Add text annotations for the DID effect
plt.annotate('DID Effect', 
             xy=(3, mean_treated_pre + (mean_control_post - mean_control_pre) + manual_did/2), 
             xytext=(3.2, mean_treated_pre + (mean_control_post - mean_control_pre) + manual_did/2),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5),
             fontsize=12)

plt.legend(['Control Trend', 'Counterfactual Trend (No Treatment)', 'Actual Treated Trend'])
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

 Line plot: mean Y by year, separate lines for treated vs control

```python
# Calculate mean Y by year and treatment status
mean_by_group = df_did.groupby(['year', 'treated'])['Y'].mean().reset_index()
mean_by_group['Group'] = mean_by_group['treated'].map({1: 'Treated', 0: 'Control'})

# Create a more visually appealing line plot
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")

# Plot the lines
sns.lineplot(data=mean_by_group, x='year', y='Y', hue='Group', 
             marker='o', markersize=10, linewidth=3,
             palette={'Treated': '#1f77b4', 'Control': '#ff7f0e'})

# Add the counterfactual trend line for treated group
treated_2023 = mean_by_group[(mean_by_group['year'] == 2023) & (mean_by_group['Group'] == 'Treated')]['Y'].values[0]
control_2023 = mean_by_group[(mean_by_group['year'] == 2023) & (mean_by_group['Group'] == 'Control')]['Y'].values[0]
control_2024 = mean_by_group[(mean_by_group['year'] == 2024) & (mean_by_group['Group'] == 'Control')]['Y'].values[0]

# Calculate the counterfactual value (what would have happened without treatment)
counterfactual_2024 = treated_2023 + (control_2024 - control_2023)

# Plot the counterfactual line
plt.plot([2023, 2024], [treated_2023, counterfactual_2024], 'r--', linewidth=2)

# Add a point for the counterfactual
plt.scatter(2024, counterfactual_2024, color='red', s=100, zorder=5, alpha=0.7)

# Annotate the treatment effect
treated_2024 = mean_by_group[(mean_by_group['year'] == 2024) & (mean_by_group['Group'] == 'Treated')]['Y'].values[0]
did_effect = treated_2024 - counterfactual_2024

plt.annotate(f'Treatment Effect\n(DiD = {did_effect:.2f})',
             xy=(2024, (treated_2024 + counterfactual_2024)/2),
             xytext=(2024.05, (treated_2024 + counterfactual_2024)/2),
             arrowprops=dict(arrowstyle='<->', color='green', lw=1.5),
             fontsize=12, color='green')

# Enhance the plot
plt.title('Mean Outcome by Year for Treated vs Control Groups', fontsize=16)
plt.xlabel('Year', fontsize=14)
plt.ylabel('Mean Outcome (Y)', fontsize=14)
plt.xticks([2023, 2024], fontsize=12)
plt.yticks(fontsize=12)

# Add a legend with custom labels
handles, labels = plt.gca().get_legend_handles_labels()
handles.append(plt.Line2D([0], [0], linestyle='--', color='red', linewidth=2))
labels.append('Counterfactual (No Treatment)')
plt.legend(handles, labels, fontsize=12, loc='upper left')

# Annotate the parallel trends assumption
plt.annotate('Parallel Trends\nAssumption', 
             xy=(2023.5, control_2023 + (control_2024 - control_2023)/2),
             xytext=(2023.5, control_2023 + (control_2024 - control_2023)/2 - 1.5),
             arrowprops=dict(arrowstyle='->', color='black', lw=1),
             fontsize=10, ha='center')

plt.tight_layout()
plt.show()

# Print the mean values for reference
print("\nMean Y by Year and Treatment Status:")
pivot_table = pd.pivot_table(df_did, values='Y', index='treated', columns='year')
pivot_table.index = ['Control', 'Treated']
pivot_table.columns = ['2023', '2024']
print(pivot_table)

# Calculate and print the DiD estimate
print("\nDifference-in-Differences Calculation:")
treated_diff = pivot_table.loc['Treated', '2024'] - pivot_table.loc['Treated', '2023']
control_diff = pivot_table.loc['Control', '2024'] - pivot_table.loc['Control', '2023']
did_estimate = treated_diff - control_diff
print(f"Treated difference (2024 - 2023): {treated_diff:.4f}")
print(f"Control difference (2024 - 2023): {control_diff:.4f}")
print(f"DiD Estimate: {did_estimate:.4f}")
```

If you had more pre-treatment years, the parallel-trends assumption would be visualized as the Treated and Control lines running consistently alongside each other almost like train tracks. A violation would look like those lines diverging or converging before the program started. For example, if the Treated stores were already "pulling away" from the Control stores in 2021 and 2022. In that case, the red dashed line would be a false counterfactual, as it would fail to account for the Treated group's pre-existing momentum.
